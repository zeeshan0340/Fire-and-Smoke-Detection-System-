```python
def check_image_label_pairs(image_dir, label_dir):

    image_extensions = {".jpg", ".jpeg", ".png", ".bmp"}

    image_stems = {
        p.stem
        for p in Path(image_dir).iterdir()
        if p.suffix.lower() in image_extensions
    }

    label_stems = {
        p.stem
        for p in Path(label_dir).glob("*.txt")
    }

    missing_labels = image_stems - label_stems
    orphan_labels = label_stems - image_stems

    return missing_labels, orphan_labels


train_missing, train_orphan = check_image_label_pairs(
    dataset_root / "train/images",
    dataset_root / "train/labels"
)

test_missing, test_orphan = check_image_label_pairs(
    dataset_root / "test/images",
    dataset_root / "test/labels"
)

print("TRAIN")
print("Missing labels:", len(train_missing))
print("Orphan labels :", len(train_orphan))

print("\nTEST")
print("Missing labels:", len(test_missing))
print("Orphan labels :", len(test_orphan))
```

```python
from collections import Counter

class_counts = Counter()

for label_file in train_labels + test_labels:

    with open(label_file, "r") as f:

        for line in f:

            line = line.strip()

            if not line:
                continue

            values = line.split()

            class_id = int(values[0])

            class_counts[class_id] += 1


print("Class distribution")
print("=" * 50)

for class_id, count in sorted(class_counts.items()):
    print(f"Class {class_id}: {count:,}")
```

```python
total_objects = sum(class_counts.values())

print("Total annotated objects:", f"{total_objects:,}")
```

```python
objects_per_image = []

for label_file in train_labels + test_labels:

    with open(label_file, "r") as f:
        lines = [
            line.strip()
            for line in f
            if line.strip()
        ]

    objects_per_image.append(len(lines))


print("Object statistics")
print("=" * 50)
print("Minimum objects/image :", min(objects_per_image))
print("Maximum objects/image :", max(objects_per_image))
print("Average objects/image :", round(np.mean(objects_per_image), 2))
print("Median objects/image  :", np.median(objects_per_image))
```

```python
import random
import shutil
from pathlib import Path

# Reproducibility
random.seed(42)

# Original dataset
original_train_images = Path("/content/D-Fire/train/images")
original_train_labels = Path("/content/D-Fire/train/labels")

original_test_images = Path("/content/D-Fire/test/images")
original_test_labels = Path("/content/D-Fire/test/labels")

# New YOLO dataset
yolo_dataset = Path("/content/D-Fire-YOLO")

train_images = yolo_dataset / "train/images"
train_labels = yolo_dataset / "train/labels"

val_images = yolo_dataset / "val/images"
val_labels = yolo_dataset / "val/labels"

test_images = yolo_dataset / "test/images"
test_labels = yolo_dataset / "test/labels"

# Create directories
for folder in [
    train_images,
    train_labels,
    val_images,
    val_labels,
    test_images,
    test_labels
]:
    folder.mkdir(parents=True, exist_ok=True)

# Get training images
images = [
    p for p in original_train_images.iterdir()
    if p.suffix.lower() in [".jpg", ".jpeg", ".png"]
]

# Shuffle reproducibly
random.shuffle(images)

# 90/10 split
split_index = int(len(images) * 0.90)

train_split = images[:split_index]
val_split = images[split_index:]

print("Original training images:", len(images))
print("New training images:", len(train_split))
print("Validation images:", len(val_split))
```

```python
def copy_images_and_labels(image_list, source_image_dir, source_label_dir,
                           destination_image_dir, destination_label_dir):

    for image_path in image_list:

        label_path = source_label_dir / f"{image_path.stem}.txt"

        if not label_path.exists():
            print("WARNING: Missing label:", image_path.name)
            continue

        shutil.copy2(
            image_path,
            destination_image_dir / image_path.name
        )

        shutil.copy2(
            label_path,
            destination_label_dir / label_path.name
        )


# Copy training set
copy_images_and_labels(
    train_split,
    original_train_images,
    original_train_labels,
    train_images,
    train_labels
)

# Copy validation set
copy_images_and_labels(
    val_split,
    original_train_images,
    original_train_labels,
    val_images,
    val_labels
)

print("Train and validation files copied successfully.")
```

```python
test_image_list = [
    p for p in original_test_images.iterdir()
    if p.suffix.lower() in [".jpg", ".jpeg", ".png"]
]

copy_images_and_labels(
    test_image_list,
    original_test_images,
    original_test_labels,
    test_images,
    test_labels
)

print("Test set copied successfully.")
```

```python
def count_files(folder, extensions=None):

    files = list(Path(folder).iterdir())

    if extensions:
        files = [
            f for f in files
            if f.suffix.lower() in extensions
        ]

    return len(files)


print("DATASET SUMMARY")
print("=" * 60)

print(
    "Train images:",
    count_files(train_images, {".jpg", ".jpeg", ".png"})
)

print(
    "Train labels:",
    count_files(train_labels, {".txt"})
)

print(
    "Val images:",
    count_files(val_images, {".jpg", ".jpeg", ".png"})
)

print(
    "Val labels:",
    count_files(val_labels, {".txt"})
)

print(
    "Test images:",
    count_files(test_images, {".jpg", ".jpeg", ".png"})
)

print(
    "Test labels:",
    count_files(test_labels, {".txt"})
)
```

```python
import cv2
import matplotlib.pyplot as plt
import random

# Class names
class_names = {
    0: "Smoke",
    1: "Fire"
}

# Select random validation image
image_path = random.choice(
    list(val_images.glob("*.jpg"))
)

label_path = val_labels / f"{image_path.stem}.txt"

# Read image
image = cv2.imread(str(image_path))
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

height, width = image.shape[:2]

# Read annotations
with open(label_path, "r") as f:

    for line in f:

        values = line.strip().split()

        if len(values) != 5:
            continue

        class_id = int(values[0])

        x_center = float(values[1]) * width
        y_center = float(values[2]) * height
        box_width = float(values[3]) * width
        box_height = float(values[4]) * height

        x1 = int(x_center - box_width / 2)
        y1 = int(y_center - box_height / 2)

        x2 = int(x_center + box_width / 2)
        y2 = int(y_center + box_height / 2)

        cv2.rectangle(
            image,
            (x1, y1),
            (x2, y2),
            (255, 0, 0),
            2
        )

        cv2.putText(
            image,
            class_names.get(class_id, f"Class {class_id}"),
            (x1, max(y1 - 10, 20)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (255, 0, 0),
            2
        )

plt.figure(figsize=(12, 8))
plt.imshow(image)
plt.axis("off")
plt.title(image_path.name)
plt.show()
```

```python
from ultralytics import YOLO
import cv2

# ============================================================
# SAFE VISION - REAL-TIME FIRE & SMOKE DETECTION
# ============================================================

MODEL_PATH = "/content/drive/MyDrive/KPK_SafeVision/models/yolo11n_fire_smoke_baseline.pt"

# Load trained model
model = YOLO(MODEL_PATH)

print("SafeVision model loaded!")
print("Classes:", model.names)

# ============================================================
# OPEN LAPTOP WEBCAM
# ============================================================

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("ERROR: Could not open webcam.")
    exit()

# Set webcam resolution
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)

print("Webcam started.")
print("Press Q to quit.")

# ============================================================
# REAL-TIME DETECTION
# ============================================================

while True:

    ret, frame = cap.read()

    if not ret:
        print("Could not read frame.")
        break

    # YOLO detection
    results = model.predict(
        source=frame,
        conf=0.25,
        imgsz=640,
        verbose=False
    )

    result = results[0]

    # --------------------------------------------------------
    # Draw detections
    # --------------------------------------------------------

    fire_detected = False
    smoke_detected = False

    for box in result.boxes:

        class_id = int(box.cls[0])
        confidence = float(box.conf[0])

        class_name = model.names[class_id].lower()

        # Bounding box coordinates
        x1, y1, x2, y2 = map(
            int,
            box.xyxy[0]
        )

        # ----------------------------------------------------
        # Detection type
        # ----------------------------------------------------

        if class_name == "fire":
            fire_detected = True

        elif class_name == "smoke":
            smoke_detected = True

        # ----------------------------------------------------
        # Draw bounding box
        # ----------------------------------------------------

        label = f"{class_name.upper()} {confidence:.2f}"

        cv2.rectangle(
            frame,
            (x1, y1),
            (x2, y2),
            (0, 0, 255),
            2
        )

        # Label background
        cv2.rectangle(
            frame,
            (x1, y1 - 30),
            (x1 + 180, y1),
            (0, 0, 255),
            -1
        )

        cv2.putText(
            frame,
            label,
            (x1 + 5, y1 - 8),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255, 255, 255),
            2
        )

    # ========================================================
    # SAFE VISION STATUS
    # ========================================================

    if fire_detected and smoke_detected:

        status = "FIRE + SMOKE DETECTED"

    elif fire_detected:

        status = "FIRE DETECTED"

    elif smoke_detected:

        status = "SMOKE DETECTED"

    else:

        status = "NO FIRE OR SMOKE"

    # ========================================================
    # DISPLAY STATUS
    # ========================================================

    cv2.putText(
        frame,
        status,
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.9,
        (0, 0, 255) if (
            fire_detected or smoke_detected
        ) else (0, 255, 0),
        3
    )

    # ========================================================
    # DISPLAY FPS
    # ========================================================

    cv2.putText(
        frame,
        "Press Q to quit",
        (20, 75),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (255, 255, 255),
        2
    )

    # Show webcam
    cv2.imshow(
        "SafeVision - Real-Time Fire & Smoke Detection",
        frame
    )

    # Quit with Q
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break


# ============================================================
# CLEANUP
# ============================================================

cap.release()
cv2.destroyAllWindows()

print("SafeVision stopped.")
```

```python
# ============================================================
# KPK SAFEVISION
# REAL-TIME FIRE & SMOKE DETECTION USING COLAB WEBCAM
# ============================================================
!pip install -q -U ultralytics
from google.colab import drive
from google.colab.output import eval_js
from IPython.display import display, Javascript, Image, update_display
from ultralytics import YOLO

import cv2
import numpy as np
import base64
import time


# ============================================================
# 1. MOUNT GOOGLE DRIVE
# ============================================================

drive.mount("/content/drive")


# ============================================================
# 2. LOAD YOUR TRAINED MODEL
# ============================================================

MODEL_PATH = (
    "/content/drive/MyDrive/"
    "KPK_SafeVision/models/"
    "yolo11n_fire_smoke_baseline.pt"
)

print("Loading SafeVision model...")

model = YOLO(MODEL_PATH)

print("Model loaded successfully!")
print("Classes:", model.names)


# ============================================================
# 3. JAVASCRIPT WEBCAM
# ============================================================

display(Javascript("""
window.safeVisionStart = async function() {

    // Remove old webcam
    const old = document.getElementById(
        "safevision-webcam"
    );

    if (old) {
        old.remove();
    }


    // Create container
    const container = document.createElement("div");

    container.id = "safevision-webcam";

    container.style.textAlign = "center";
    container.style.margin = "10px";


    // Create video
    const video = document.createElement("video");

    video.width = 640;
    video.height = 480;

    video.autoplay = true;
    video.playsInline = true;

    video.style.border = "3px solid green";
    video.style.borderRadius = "10px";


    container.appendChild(video);

    document.body.appendChild(container);


    // Request camera
    const stream =
        await navigator.mediaDevices.getUserMedia({
            video: {
                width: 640,
                height: 480
            },
            audio: false
        });


    video.srcObject = stream;

    await video.play();


    // Save globally
    window.safeVisionVideo = video;
    window.safeVisionStream = stream;


    return "SafeVision webcam started";
};


window.safeVisionGetFrame = async function() {

    const video = window.safeVisionVideo;


    if (!video) {

        throw new Error(
            "SafeVision webcam is not running."
        );

    }


    // Make sure video is ready
    if (video.readyState < 2) {

        return null;

    }


    // Create canvas
    const canvas =
        document.createElement("canvas");


    canvas.width =
        video.videoWidth;

    canvas.height =
        video.videoHeight;


    // Draw current webcam frame
    const context =
        canvas.getContext("2d");


    context.drawImage(
        video,
        0,
        0,
        canvas.width,
        canvas.height
    );


    // Convert to JPEG
    return canvas.toDataURL(
        "image/jpeg",
        0.70
    );
};


window.safeVisionStop = function() {

    if (window.safeVisionStream) {

        window.safeVisionStream
            .getTracks()
            .forEach(
                track => track.stop()
            );

    }

    const container =
        document.getElementById(
            "safevision-webcam"
        );

    if (container) {
        container.remove();
    }

    window.safeVisionVideo = null;
    window.safeVisionStream = null;

    return "SafeVision webcam stopped";
};
"""))


# ============================================================
# 4. START WEBCAM
# ============================================================

print(
    eval_js(
        "safeVisionStart()"
    )
)

print()
print("=" * 60)
print("KPK SAFEVISION REAL-TIME DETECTION")
print("=" * 60)
print()
print("Webcam is running.")
print()
print("Allow camera permission if your browser asks.")
print()
print("To stop detection:")
print("Press the STOP/INTERRUPT button of this Colab cell.")
print()


# ============================================================
# 5. CREATE OUTPUT DISPLAY
# ============================================================

display_id = "safevision-detection-output"


# ============================================================
# 6. FPS VARIABLES
# ============================================================

frame_count = 0

start_time = time.time()


# ============================================================
# 7. REAL-TIME DETECTION LOOP
# ============================================================

try:

    while True:

        # ====================================================
        # GET CURRENT WEBCAM FRAME
        # ====================================================

        data = eval_js(
            "safeVisionGetFrame()"
        )


        # If frame isn't ready yet
        if data is None:

            time.sleep(0.01)

            continue


        # ====================================================
        # DECODE FRAME
        # ====================================================

        encoded = data.split(",")[1]

        image_bytes = base64.b64decode(
            encoded
        )


        image_array = np.frombuffer(
            image_bytes,
            dtype=np.uint8
        )


        frame = cv2.imdecode(
            image_array,
            cv2.IMREAD_COLOR
        )


        if frame is None:

            continue


        # ====================================================
        # YOLO DETECTION
        # ====================================================

        results = model.predict(

            source=frame,

            conf=0.25,

            imgsz=640,

            device='cpu',

            verbose=False

        )


        result = results[0]


        # ====================================================
        # DETECTION FLAGS
        # ====================================================

        fire_detected = False

        smoke_detected = False


        fire_count = 0

        smoke_count = 0


        # ====================================================
        # DRAW BOUNDING BOXES
        # ====================================================

        for box in result.boxes:


            # -----------------------------------------------
            # Class
            # -----------------------------------------------

            class_id = int(
                box.cls[0]
            )


            # -----------------------------------------------
            # Confidence
            # -----------------------------------------------

            confidence = float(
                box.conf[0]
            )


            # -----------------------------------------------
            # Class name
            # -----------------------------------------------

            class_name = (
                model.names[class_id]
                .lower()
            )


            # -----------------------------------------------
            # Coordinates
            # -----------------------------------------------

            x1, y1, x2, y2 = map(
                int,
                box.xyxy[0]
            )


            # -----------------------------------------------
            # FIRE
            # -----------------------------------------------

            if class_name == "fire":

                fire_detected = True

                fire_count += 1


            # -----------------------------------------------
            # SMOKE
            # -----------------------------------------------

            elif class_name == "smoke":

                smoke_detected = True

                smoke_count += 1


            # -----------------------------------------------
            # Label
            # -----------------------------------------------

            label = (
                f"{class_name.upper()} "
                f"{confidence * 100:.1f}%"
            )


            # -----------------------------------------------
            # Bounding box
            # -----------------------------------------------

            cv2.rectangle(

                frame,

                (x1, y1),

                (x2, y2),

                (0, 0, 255),

                3

            )


            # -----------------------------------------------
            # Label background
            # -----------------------------------------------

            label_width = 190

            label_height = 32


            label_y1 = max(
                y1 - label_height,
                0
            )


            label_y2 = y1


            cv2.rectangle(

                frame,

                (x1, label_y1),

                (
                    x1 + label_width,
                    label_y2
                ),

                (0, 0, 255),

                -1

            )


            # -----------------------------------------------
            # Label text
            # -----------------------------------------------

            cv2.putText(

                frame,

                label,

                (
                    x1 + 5,
                    label_y2 - 8
                ),

                cv2.FONT_HERSHEY_SIMPLEX,

                0.6,

                (255, 255, 255),

                2

            )


        # ====================================================
        # DETERMINE STATUS
        # ====================================================

        if fire_detected and smoke_detected:

            status = "FIRE + SMOKE DETECTED"

            status_color = (
                0,
                0,
                255
            )


        elif fire_detected:

            status = "FIRE DETECTED"

            status_color = (
                0,
                0,
                255
            )


        elif smoke_detected:

            status = "SMOKE DETECTED"

            status_color = (
                0,
                165,
                255
            )


        else:

            status = "NO FIRE OR SMOKE Detected"

            status_color = (
                0,
                200,
                0
            )


        # ====================================================
        # STATUS PANEL
        # ====================================================

        cv2.rectangle(

            frame,

            (10, 10),

            (550, 70),

            (0, 0, 0),

            -1

        )


        # ====================================================
        # STATUS TEXT
        # ====================================================

        cv2.putText(

            frame,

            status,

            (20, 50),

            cv2.FONT_HERSHEY_SIMPLEX,

            0.85,

            status_color,

            3

        )


        # ====================================================
        # COUNTERS
        # ====================================================

        counter_text = (
            f"Fire: {fire_count}   "
            f"Smoke: {smoke_count}"
        )


        cv2.putText(

            frame,

            counter_text,

            (15, 105),

            cv2.FONT_HERSHEY_SIMPLEX,

            0.6,

            (255, 255, 255),

            2

        )


        # ====================================================
        # FPS
        # ====================================================

        frame_count += 1


        elapsed = (
            time.time() - start_time
        )


        fps = (
            frame_count / elapsed
            if elapsed > 0
            else 0
        )


        cv2.putText(

            frame,

            f"FPS: {fps:.1f}",

            (15, 135),

            cv2.FONT_HERSHEY_SIMPLEX,

            0.6,

            (255, 255, 255),

            2

        )


        # ====================================================
        # CONVERT FRAME TO JPEG
        # ====================================================

        success, buffer = cv2.imencode(

            ".jpg",

            frame,

            [
                cv2.IMWRITE_JPEG_QUALITY,
                80
            ]

        )


        if not success:

            continue


        # ====================================================
        # UPDATE DISPLAY
        # ====================================================

        update_display(

            Image(
                data=buffer.tobytes()
            ),

            display_id=display_id

        )


        # ====================================================
        # SMALL DELAY
        # ====================================================

        time.sleep(0.01)


# ============================================================
# STOP / CLEANUP
# ============================================================

except KeyboardInterrupt:

    print()
    print("Stopping SafeVision...")


finally:

    try:

        eval_js(
            "safeVisionStop()"
        )

    except:

        pass


    print()
    print("=" * 60)
    print("SAFЕVISION STOPPED")
    print("=" * 60)
```
